```python
def to_gray(img):
  weights = np.array([0.299, 0.587, 0.114])
  return img @ weights
gray = to_gray(img)
plt.figure(figsize=(8,5))
plt.imshow(gray, cmap="gray")
plt.title("haqiqiy oq-qora(kulrang ) rasm ")
plt.axis("off")
plt.show()
```

```python
def adjust_brightness(img, delta):
  natija=img.astype(np.float64)+delta
  natija=np.clip(natija, 0, 255)
  return natija.astype(np.uint8)
img1 = adjust_brightness(img , 50)
plt.figure(figsize=(8,5))
plt.subplot(121)
plt.imshow(img1)
plt.title("Brightness")
plt.axis("off")
img2 = adjust_brightness(img , -50)
plt.subplot(122)
plt.imshow(img2)
plt.title("Brightness")
plt.axis("off")
plt.show()
```

```python
def invert(img):
  return 255-img
negativ_img= invert(img)
plt.figure(figsize=(16,24))
plt.subplot(121)
plt.imshow(negativ_img)
plt.title("Negativ rasm")
plt.axis("off")
plt.subplot(122)
plt.imshow(img)
plt.title("Orginal rasm")
plt.axis("off")
plt.show()
```

```python
def adjust_contrast(img, factor):
  natija = 128+ (img.astype(np.float64)-128)*factor
  natija=np.clip(natija, 0, 255)
  return natija.astype(np.uint8)
img_yuqori = adjust_contrast(img, 1.8)
img_past = adjust_contrast(img, 0.5)
plt.figure(figsize=(16,24))
plt.subplot(121)
plt.imshow(img_yuqori)
plt.title("Yuqori kontrast (factor=1.8)")
plt.axis("off")
plt.subplot(122)
plt.imshow(img_past)
plt.title("Past kontrast (factor=0.5)")
plt.axis("off")
plt.show()

```
